```python
%matplotlib inline
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# If you saved the previous script as analyze_block_confusion.py, import functions:
from analyze_block_confusion import (
    _load_file,
    find_source_lightcurve,
    run_confusion_experiment,
    plot_confusion_matrix,
    plot_efficiency_vs_bins,
    plot_detected_hist
)
```

```python
# Option A: load a wtlike file and extract a particular source
# Replace with your local path and source key
wtlike_path = Path("/home/bean625/agnsteps_copy/files/source_info_v3.pkl")   # e.g. "./wtlike_outputs/sim_realization.pkl"
source_key = "J0000.3+6052"                             # use the actual source key in your file

use_synthetic = False  # set True to use synthetic baseline instead of reading a file

if not use_synthetic:
    try:
        obj = _load_file(wtlike_path)
        lc = find_source_lightcurve(obj, source_key)
        base_flux = np.asarray(lc.get("flux", []), dtype=float)
        base_errors = lc.get("errors", None)
        if base_errors is not None:
            base_errors = np.asarray(base_errors, dtype=float)
            if base_errors.shape != base_flux.shape:
                base_errors = None
        print(f"Loaded light curve for {source_key}: {len(base_flux)} bins")
    except Exception as e:
        print("Failed to load wtlike file or extract source:", e)
        print("Falling back to synthetic baseline.")
        use_synthetic = True

# Option B: synthetic baseline (flat flux with small jitter)
if use_synthetic:
    n_bins = 100
    rng = np.random.default_rng(1)
    base_flux = np.ones(n_bins) * rng.normal(1.0, 0.02, size=n_bins)
    base_errors = np.maximum(1e-6, np.abs(base_flux) * 0.05)
    print(f"Using synthetic baseline: {n_bins} bins")
```

```python
# Experiment parameters (tweak as desired)
n_trials = 500                   # trials per (n_steps, step_size)
step_sizes = [1.5, 2.0, 3.0]     # multiplicative step amplitudes
n_steps_list = [1, 2]            # inject S steps -> true blocks = S+1
min_bins_per_block = 5           # enforce minimum block width
p0 = 0.05                        # significance threshold for z-score detector
rng_seed = 42                    # reproducibility
detector = "zscore"              # "zscore" or "astropy" (if available)

# Run experiment (this will return trial-level DataFrame and a summary dict)
df_trials, summary = run_confusion_experiment(
    base_flux=base_flux,
    base_errors=base_errors,
    n_trials=n_trials,
    step_sizes=step_sizes,
    n_steps_list=n_steps_list,
    rng_seed=rng_seed,
    min_bins_per_block=min_bins_per_block,
    p0=p0,
    detector=detector
)

# Quick look at the produced data
df_trials.head()
```

```python
# Print confusion counts and fractions
print("Confusion matrix (counts):")
print(summary["confusion_counts"])
print("\nConfusion matrix (fractions, rows = true blocks):")
print(summary["confusion_fractions"])

# Detection efficiency summary (true blocks -> step_size -> eff and 95% CI)
print("\nDetection efficiency summary:")
for true_b, bystep in sorted(summary["detection_efficiency"].items()):
    print(f"\nTrue blocks = {true_b}:")
    for step, statsd in sorted(bystep.items()):
        eff = statsd["eff"]
        lo = statsd["ci_lo"]
        hi = statsd["ci_hi"]
        n = statsd["n"]
        print(f"  step={step:.2f}: eff={eff:.1%} 95% CI=({lo:.1%}, {hi:.1%})  n={n}")

# Save trial-level CSV if desired
out_csv = "block_confusion_trials_example.csv"
df_trials.to_csv(out_csv, index=False)
print(f"\nSaved trial-level results to {out_csv}")
```
