```python
from wtlike import WtLike
from pylib.mc_dev import BBsim, check_step, p_view
from utilities.ipynb_docgen import show
import matplotlib.pyplot as plt
import pandas as pd
from IPython.utils.capture import capture_output

# Adjust these parameters and rerun this cell.
sweep_source_name = name
rng = np.random.default_rng(seed = 62508)
bin_widths_days = np.round(rng.normal(loc=30, scale=10, size=1000).clip(5, 100))
num_partitions = 100
nproc = 8
p0 = 0.05
partition_delta_days = 1500
reverse = True

sweep_results = []
sweep_samples = []

for width_index, bin_width_days in enumerate(bin_widths_days, start=1):
    try:
        with capture_output():
            sweep_wtl = WtLike(
                sweep_source_name,
                time_bins=(0, 0, bin_width_days),
                clear=False,
            )

            sweep_bb = sweep_wtl.bb_view(p0=p0)
            sweep_step = check_step(sweep_bb, showit=False)

            sweep_range = (
                sweep_step.time - partition_delta_days,
                sweep_step.time + partition_delta_days,
            )
            sweep_partition = p_view(sweep_wtl, *sweep_range)

            _, simulation_results = BBsim.run_many(
                sweep_partition,
                seeds=range(num_partitions),
                nproc=nproc,
                step_time=sweep_step.time,
            )

            sweep_samples.append(
                simulation_results.assign(
                    bin_width_days=bin_width_days,
                    true_step_time=sweep_step.time,
                )
            )
            sweep_results.append({
                "bin_width_days": bin_width_days,
                "input_cells": len(sweep_wtl.cells),
                "partition_cells": len(sweep_partition.cells),
                "mean_blocks": simulation_results.nbb.mean(),
                "mean_step_time": simulation_results.time.mean(),
                "mean_ts": simulation_results.ts.mean(),
                "mean_ratio": simulation_results.ratio.mean(),
            })
        print(f"{bin_width_days}d: done")

    except ValueError as e:
        print(f"{bin_width_days}d: skipped ({' '.join(str(e).split())})")
        continue
    except KeyboardInterrupt:
        print(f"{bin_width_days}d: interrupted")
        break
    except Exception as e:
        print(f"{bin_width_days}d: failed ({type(e).__name__}: {' '.join(str(e).split())})")
        continue

sweep_summary = pd.DataFrame(sweep_results)
display(sweep_summary)

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)
metrics = [
    ("mean_blocks", "Average number of blocks"),
    ("mean_step_time", "Average detected step time (MJD)"),
    ("mean_ts", "Average variability TS"),
    ("mean_ratio", "Average step ratio"),
]
for ax, (column, ylabel) in zip(axes.flat, metrics):
    ax.plot(sweep_summary.bin_width_days, sweep_summary[column], "o-")
    ax.set(xlabel="Bin width (days)", ylabel=ylabel)
    ax.grid(alpha=0.3)
fig.suptitle(
    f"Averages over {num_partitions:,} Bayesian-block partitions per bin width"
)
fig.tight_layout()
show(fig)
```

```python
if not sweep_samples:
    raise ValueError("Cell 7 did not produce any simulation results to average.")

all_simulations = pd.concat(sweep_samples, ignore_index=True)
average_results = all_simulations[["nbb", "time", "ts", "ratio"]].mean()
print("Average results across all simulations:")
display(average_results.rename({
    "nbb": "Average number of blocks",
    "time": "Average detected step time (MJD)",
    "ts": "Average variability TS",
    "ratio": "Average step ratio",
}).to_frame(name="mean").round(3))

# Overlay the ensemble-average step on the observed light curve from cell 6.
bb_full.plot(log=True)
ax = plt.gca()
fits = bb_full.fits
start_mjd = float(fits.t.iloc[0] - fits.tw.iloc[0] / 2)
end_mjd = float(fits.t.iloc[-1] + fits.tw.iloc[-1] / 2)
mean_step_time = float(average_results["time"])
mean_ratio = float(average_results["ratio"])
baseline_flux = float(fits.fit.iloc[0].flux)

ax.step(
    [start_mjd, mean_step_time, end_mjd],
    [baseline_flux, baseline_flux * mean_ratio, baseline_flux * mean_ratio],
    where="pre",
    color="darkorange",
    linestyle="--",
    linewidth=2,
    label="Ensemble-average simulated step",
)
ax.axvline(mean_step_time, color="darkorange", linestyle=":", alpha=0.8)
ax.legend()
fig = plt.gcf()
fig.suptitle("Observed light curve with ensemble-average simulated step")
show(fig)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from pylib.mc_dev import TS

# Re-run Bayesian Blocks with penalties from 0.5% through 10%.
penalties = np.arange(0.005, 0.1001, 0.005)
penalty_results = []

for penalty in penalties:
    bb = wtl.bb_view(p0=float(penalty))
    fits = bb.fits
    penalty_results.append({
        "penalty_percent": penalty * 100,
        "blocks": len(fits),
        "tsvar": float(TS(fits)),
        "step_ratio": float(fits.fit.iloc[-1].flux / fits.fit.iloc[0].flux),
        "step_mjd": float(fits.t.iloc[0] + fits.tw.iloc[0] / 2),
    })

penalty_results = pd.DataFrame(penalty_results)
print("Bayesian Blocks results by penalty:")
display(penalty_results.round(3))

averaged_results = penalty_results[["blocks", "tsvar", "step_ratio", "step_mjd"]].mean()
print("\nMean results across penalty settings:")
display(averaged_results.rename("mean").to_frame().round(3))

fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
axes[0].plot(penalty_results.penalty_percent, penalty_results.blocks, marker="o")
axes[0].axhline(penalty_results.blocks.mean(), color="gray", linestyle="--", label="Mean")
axes[0].set_ylabel("Number of blocks")
axes[0].legend()
axes[1].plot(penalty_results.penalty_percent, penalty_results.tsvar, marker="o", color="darkorange")
axes[1].axhline(penalty_results.tsvar.mean(), color="gray", linestyle="--", label="Mean")
axes[1].set(xlabel="Penalty (%)", ylabel="TSvar")
axes[1].legend()
fig.suptitle("Bayesian Blocks sensitivity to penalty")
fig.tight_layout()
plt.show()
```

# Interval Testing To Estimate Accuracy #interval-testing #bayesian-blocks #monte-carlo #binning
The goal of this is to run multiple simulations with different bin widths averaging at the bin width we are using in experimentation. This is being done in order to determine the accuracy of Bayesian-block counting within that bin width.

### Multiple Simulations
#### Values
In this instance, I created 10 test values that averaged to 30.1 (the bin width used for actual testing). In a larger simulation with a better processor, I would run 1000 test values that are randomly selected to average on 30. But since I do not have that and want replicability, we are using a small pool of hand-selected values
#### Simulation
These values are run through the exact same simulation and partitioning as the 30-day bins are. The values are stored and the simulation moves onto the next bin width. Each run of binning takes approximately 1.5 minutes to complete on this computer.
#### Data Analysis
Once all the test values have been run, the numbers of blocks, variability, and step time are all both graphed individually and averaged into whole datasets.

### Future Plans
My goal is to run actual statistical analysis on the results from the simulations and from plain 30-day bins to see if they are reasonably close. Further research will be needed to determine the best statistical test.

```python
# Run many independent simulations. Do not use cache_file.
N = 100
nproc = 8

bbs, df_all = BBsim.run_many(
    partition_view,
    seeds=range(N),
    nproc=nproc,
    step_time=step.time,
)

# Collect the Bayesian-block cells from every simulation.
block_frames = []

for simulation_id, bb in enumerate(bbs):
    cells = bb.cells.copy()
    cells.insert(0, "simulation", simulation_id)
    cells.insert(1, "block", np.arange(len(cells)))
    block_frames.append(cells.reset_index(drop=True))

blocks_df = pd.concat(block_frames, ignore_index=True)

# Ensure chronological ordering within each simulation.
time_column = next(
    (c for c in ("time", "t", "mjd", "tstart") if c in blocks_df.columns),
    None,
)

if time_column is not None:
    blocks_df = (
        blocks_df
        .sort_values(["simulation", time_column])
        .reset_index(drop=True)
    )

# Describe the diversity of outcomes.
outcome_df = (
    blocks_df.groupby("simulation")
    .size()
    .rename("num_blocks")
    .reset_index()
)

display(outcome_df.describe())
display(blocks_df.head())
```
